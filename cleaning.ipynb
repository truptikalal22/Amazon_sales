```python
import pandas as pd
```

```python
df=pd.read_csv("Amazon_sales.csv")
```

```python
print("Original shape:",df.shape)
print("\n missing values \n " , df.isnull().sum())
print("\n duplicates value\n",df.duplicated())
```

```python
df = df.drop_duplicates()
print("after removing duplicate",df.shape)
```

```python
df = df.drop(columns=["New", "PendingS", "index"])
print("column now:",df.columns.tolist())
```

```python
print("\n missing values \n " , df.isnull().sum())
```

```python
#Remove rows where Amount or currency is missing
df = df.dropna(subset=["Amount","currency"])

#missing value fill with "unkown"
df["ship-city"] = df["ship-city"].fillna("unkown")
df["ship-state"] = df["ship-state"].fillna("unkown")
df["ship-postal-code"] = df["ship-postal-code"].fillna(0)


df["fulfilled-by"] = df["fulfilled-by"].fillna("unkown")
```

```python
df["Date"] = pd.to_datetime(df["Date"])
print("Date type is now:", df["Date"].dtype)
```

```python
df.to_csv("Amazon_sales_cleaned.csv", index=False)
print("Done! Cleaned file saved as Amazon_sales_cleaned.csv")
print("Final shape:", df.shape)
```
